```python
dataset = fetch_ucirepo(id=601)
X = dataset.data.features.copy()
target_df = dataset.data.targets.copy()
y = target_df["Machine failure"].astype(int)

# Remove identifiers and failure-mode labels to avoid target leakage.
X = X.drop(columns=["UDI", "UID", "Product ID", "TWF", "HDF", "PWF", "OSF", "RNF", "Machine failure"], errors="ignore")
print("Feature shape:", X.shape)
print("Target distribution:\n", y.value_counts())
display(X.head())
```

```python
display(X.describe(include="all").T)
plt.figure(figsize=(5,4))
sns.countplot(x=y)
plt.title("Machine failure class distribution")
plt.xlabel("Machine failure (0 = No, 1 = Yes)")
plt.show()

numeric_cols = X.select_dtypes(include=np.number).columns
X[numeric_cols].hist(figsize=(12,8), bins=25)
plt.tight_layout()
plt.show()
```

```python
categorical_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()
numeric_cols = [c for c in X.columns if c not in categorical_cols]

preprocessor = ColumnTransformer([
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median")),
                      ("scaler", StandardScaler())]), numeric_cols),
    ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                      ("onehot", OneHotEncoder(handle_unknown="ignore"))]), categorical_cols)
])

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=RANDOM_STATE
)
print("Train:", X_train.shape, "Test:", X_test.shape)
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=2000, class_weight="balanced"),
    "Decision Tree": DecisionTreeClassifier(max_depth=8, min_samples_leaf=5, class_weight="balanced", random_state=RANDOM_STATE),
    "Random Forest": RandomForestClassifier(n_estimators=350, min_samples_leaf=2, class_weight="balanced_subsample", random_state=RANDOM_STATE, n_jobs=-1),
    "Extra Trees": ExtraTreesClassifier(n_estimators=350, min_samples_leaf=2, class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1)
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
pipelines, rows = {}, []
for name, estimator in models.items():
    pipe = ImbPipeline([
        ("preprocess", preprocessor),
        ("smote", SMOTE(random_state=RANDOM_STATE, k_neighbors=3)),
        ("classifier", estimator)
    ])
    scores = cross_validate(pipe, X_train, y_train, cv=cv,
        scoring={"precision":"precision", "recall":"recall", "f1":"f1", "roc_auc":"roc_auc"},
        n_jobs=-1)
    pipelines[name] = pipe
    rows.append({"Model":name, "Precision":scores["test_precision"].mean(),
                 "Recall":scores["test_recall"].mean(), "F1":scores["test_f1"].mean(),
                 "ROC-AUC":scores["test_roc_auc"].mean()})
comparison = pd.DataFrame(rows).sort_values("F1", ascending=False)
display(comparison)
```

```python
best_name = comparison.iloc[0]["Model"]
best_model = pipelines[best_name]
best_model.fit(X_train, y_train)
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

print("Selected model:", best_name)
print(classification_report(y_test, y_pred, zero_division=0))
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Failure precision:", precision_score(y_test, y_pred, zero_division=0))
print("Failure recall:", recall_score(y_test, y_pred, zero_division=0))
print("Failure F1:", f1_score(y_test, y_pred, zero_division=0))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
print("Average precision:", average_precision_score(y_test, y_prob))

ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["No failure", "Failure"], values_format="d")
plt.title(f"Confusion Matrix — {best_name}")
plt.show()
```

```python
out_dir = pathlib.Path("outputs")
model_dir = pathlib.Path("models")
out_dir.mkdir(exist_ok=True)
model_dir.mkdir(exist_ok=True)
joblib.dump(best_model, model_dir/"machine_failure_pipeline.joblib")
joblib.dump(list(X.columns), model_dir/"feature_columns.joblib")
comparison.to_csv(out_dir/"model_comparison.csv", index=False)
metrics = {
    "selected_model": best_name,
    "accuracy": float(accuracy_score(y_test, y_pred)),
    "precision_failure": float(precision_score(y_test, y_pred, zero_division=0)),
    "recall_failure": float(recall_score(y_test, y_pred, zero_division=0)),
    "f1_failure": float(f1_score(y_test, y_pred, zero_division=0)),
    "roc_auc": float(roc_auc_score(y_test, y_prob)),
    "average_precision": float(average_precision_score(y_test, y_prob))
}
(out_dir/"metrics.json").write_text(json.dumps(metrics, indent=2))
print(metrics)
```
